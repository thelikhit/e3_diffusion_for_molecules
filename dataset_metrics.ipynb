```python
import matplotlib.pyplot as plt
import numpy as np

# ---------- Data ----------
n_nodes = {
    22: 3393, 17: 13025, 23: 4848, 21: 9970, 19: 13832, 20: 9482,
    16: 10644, 13: 3060, 15: 7796, 25: 1506, 18: 13364, 12: 1689,
    11: 807, 24: 539, 14: 5136, 26: 48, 7: 16, 10: 362, 8: 49,
    9: 124, 27: 266, 4: 4, 29: 25, 6: 9, 5: 5, 3: 1
}

atom_types = {  # id → count
    1: 635559, 2: 101476, 0: 923537, 3: 140202, 4: 2323
}

# Encoders / decoders
atom_decoder = ['H', 'C', 'N', 'O', 'F']   # index → symbol

# ---------- Probability normalization ----------
def to_probability(arr):
    return arr / arr.sum()

# ---------- Plot 1: Molecule size distribution ----------
sizes = np.array(list(n_nodes.keys()))
counts = np.array(list(n_nodes.values()))
sorted_idx = np.argsort(sizes)

sizes_sorted = sizes[sorted_idx]
prob_sorted = to_probability(counts[sorted_idx])

plt.figure(figsize=(12, 5))
plt.bar(sizes_sorted, prob_sorted)
plt.xlabel("Number of atoms per molecule")
plt.ylabel("Normalized Frequency")
plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()

# ---------- Plot 2: Atom type distribution (with labels) ----------
type_ids = np.array(list(atom_types.keys()))
type_counts = np.array(list(atom_types.values()))
type_probs = to_probability(type_counts)

# Convert type IDs to element symbols
labels = [atom_decoder[i] for i in type_ids]

plt.figure(figsize=(8, 5))
plt.bar(labels, type_probs)   # labels on x-axis
plt.xlabel("Atom Type")
plt.ylabel("Normalized Frequency")
plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```
