```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([9, 12, 16, 20, 25])
y = np.array([10, 8, 4, 2, 1.4])
```

```python
coeff_1 = np.polyfit(x, y, 1)
coeff_2 = np.polyfit(x, y, 2)
coeff_3 = np.polyfit(x, y, 3)
```

```python
x_ = np.arange(3, 30)
y_pred_1 = np.polyval(coeff_1, x_)
y_pred_2 = np.polyval(coeff_2, x_)
y_pred_3 = np.polyval(coeff_3, x_)
```
